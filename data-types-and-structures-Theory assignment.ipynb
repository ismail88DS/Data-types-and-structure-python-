## 16. What is a nested dictionary, and give an example of its use case  
**Answer:**  
A nested dictionary is a dictionary that contains another dictionary as its value. It is useful for representing hierarchical or structured data.

**Example:**
```python

    "Sara": {"Department": "HR", "Age": 28}

## 17. Describe the time complexity of accessing elements in a dictionary  
**Answer:**  
Accessing elements in a dictionary has an average time complexity of **O(1)**. This is because dictionaries use a hash table internally, allowing constant-time retrieval of values using their keys. In rare cases of hash collisions, the time complexity can degrade to **O(n)**, but this is uncommon in practice.

## 19. Why are dictionaries considered unordered, and how does that affect data retrieval  
**Answer:**  
Prior to Python 3.7, dictionaries were considered unordered because they did not preserve the order in which items were inserted. This meant that when retrieving data, the sequence of key-value pairs could vary. Starting with Python 3.7, dictionaries preserve insertion order, but they are still accessed by key rather than position. This affects retrieval when the order of data is important, as dictionaries are optimized for fast key-based access, not sequential operations.

## 20. Explain the difference between a list and a dictionary in terms of data retrieval  
**Answer:**  
- **List**: Elements are accessed by their index position. Useful for ordered collections where position matters.  
- **Dictionary**: Elements are accessed by their keys. Ideal for labeled data where each item has a unique identifier.

**Example:**
```python
# List
colors = ["red", "green", "blue"]
print(colors[1])  # Output: green

# Dictionary
color_codes = {"red": "#FF0000", "green": "#00FF00"}
print(color_codes["green"])  # Output: #00FF00
